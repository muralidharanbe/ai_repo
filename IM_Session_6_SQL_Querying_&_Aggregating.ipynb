```python
import sqlite3 as s
conn=s.connect("orders.db")
cur=conn.cursor()
print("Data Base Connectedd Successfully")

```

```python
cur.execute("DROP TABLE IF EXISTS orders")
cur.execute("DROP TABLE IF EXISTS restaurants")

cur.execute("""CREATE TABLE restaurants
(restaurant_id INTEGER PRIMRY KEY,
 name TEXT NOT NULL,
 cuisine TEXT NOT NULL)"""
 )

cur.execute("""CREATE TABLE orders
(order_id INTEGER PRIMRY KEY,
 customer TEXT NOT NULL,
 restaurant TEXT NOT NULL,
 items INTEGER NOT NULL,
 amount  REAL NOT NULL,
 status TEXT NOY NULL)"""
 )

conn.commit()
print("Tables Created Successfully")
```

```python
orders = [
    (101, "Anita", "Spice Route", 3, 540.0, "delivered"),
    (102, "Ravi", "Biryani House", 2, 380.0, "delivered"),
    (103, "Meena", "Dosa Corner", 4, 260.0, "cancelled"),
    (104, "Anita", "Dosa Corner", 1, 120.0, "delivered"),
    (105, "Karthik", "Spice Route", 5, 910.0, "delivered"),
    (106, "Divya", "Biryani House", 2, 450.0, "delivered"),
    (107, "Ravi", "Pizza Point", 3, 690.0, "cancelled"),
    (108, "Suresh", "Dosa Corner", 2, 300.0, "delivered")
]
restaurants = [
    (1, "Spice Route", "North Indian"),
    (2, "Biryani House", "Biryani"),
    (3, "Dosa Corner", "South Indian"),
    (4, "Pizza Point", "Italian")
]
cur.executemany(
    "INSERT INTO orders VALUES (?, ?, ?, ?, ?, ?)",
    orders
)
cur.executemany(
    "INSERT INTO restaurants VALUES (?, ?, ?)",
    restaurants
)
conn.commit()
print("Sample data inserted successfully!")
```

```python
order_data=cur.execute("SELECT * FROM orders")
print("\nOrder Table")
for row in order_data:
    print(row)

restaurant_data=cur.execute("SELECT * FROM restaurants")
print("\nRestaurant Table")
for row in restaurant_data:
    print(row)
```

```python
order_data=cur.execute("SELECT customer, amount FROM orders ORDER BY amount DESC")
print("\nOrder Table")
for row in order_data:
    print(row)
```

```python
order_data=cur.execute("SELECT restaurant, amount FROM orders ORDER BY restaurant DESC")
print("\nOrder Table")
for row in order_data:
    print(row)
```

```python
order_data=cur.execute("SELECT customer, status, amount FROM orders ORDER BY status ASC, amount DESC")
print("\nOrder Table")
for row in order_data:
    print(row)
```

```python
order_data=cur.execute("SELECT restaurant, COUNT(*) AS num_orders FROM orders GROUP BY restaurant")
print("\nOrder Table")
for row in order_data:
    print(row)
```

```python
order_data=cur.execute("SELECT restaurant, SUM(amount) AS total_amount FROM orders GROUP BY restaurant")
print("\nOrder Table")
for row in order_data:
    print(row)
```

```python
import sqlite3

conn = sqlite3.connect("orders.db")
cur = conn.cursor()

try:
  cur.execute("""INSERT INTO orders(order_id,customer,restaurant,items,status)
  VALUES (109,'Priya','Dosa Corner',2,'Delivered')""")
  conn.commit()
except sqlite3.IntegrityError as error:
  print("Databse rejected the Order",error)
except sqlite3.OperationalError as error:
    print("Database problem:", error)
```

```python
for row in cur.execute("SELECT customer,amount FROM orders"):
  print(row)
```

```python
for row in cur.execute("SELECT * FROM orders"):
  print(row)
```

```python
for row in cur.execute("SELECT * FROM orders where status='cancelled'"):
  print(row)
```

```python
for row in cur.execute("SELECT * FROM orders where amount>500"):
  print(row)
```

```python
for row in cur.execute("SELECT order_id,customer,amount,status FROM orders where amount>=300 and status='delivered'"):
  print(row)
```

```python
for row in cur.execute("SELECT order_id,customer,amount,status FROM orders where amount>=300 or status='delivered'"):
  print(row)
```

```python
order_data=cur.execute("""SELECT restaurant, COUNT(*) AS num_orders FROM orders
                       GROUP BY restaurant
                       HAVING count(*)>1""")
print("\nOrder Table")
for row in order_data:
    print(row)
```
